```python
#8.CLEANING DATASET AND CREATING USEFUL COLUMNS 
#8.1- DROPPING DUPLICATE RECORDS
df1=df1.drop_duplicates().copy()
print(" 8.1: Dropping done successfully")

#8.2 CREATING NEW USEFUL COLUMNS 
#8.1-a) - CREATING FLIGHT DATE COLUMN 
df1["FLIGHT_DATE"]=pd.to_datetime(df1[["YEAR","MONTH","DAY"]])

print("8.2 -a)FLIGHT DATES:")
print(df1["FLIGHT_DATE"])

#8.2-b)CREATING ROUTE COLUMN
df1["ROUTE"] = df1["ORIGIN_AIRPORT"].astype(str) + " - " + df1["DESTINATION_AIRPORT"].astype(str)

print(df1[["ORIGIN_AIRPORT", "DESTINATION_AIRPORT", "ROUTE"]].head())
print("8.2-b) ROUTES:")
print(df1["ROUTE"])

#8.3-c)CREATING DEPARTURE & ARRIVAL DELAY COLUMNS (GREATURE THAN 15 MIN)
print("DEPARTURE DELAY TIME:")
df1["DEPARTURE_DELAYED"]=df1["DEPARTURE_DELAY"]>15
print(df1["DEPARTURE_DELAYED"])

print("ARRIVAL DEALY TIME:")
df1["ARRIVAL_DELAYED"]=df1["ARRIVAL_DELAY"]>15
print(df1["ARRIVAL_DELAYED"])
```

```python
#9.ADDING AIRLINE NAMES 
# Adding airline names to the flights dataset

df3 = df3.rename(columns={"AIRLINE": "AIRLINE_NAME"})

df1 = df1.merge(
    df3[["IATA_CODE", "AIRLINE_NAME"]],
    left_on="AIRLINE",
    right_on="IATA_CODE",
    how="left"
)

df1 = df1.drop(columns=["IATA_CODE"])

# Check the result
print(df1[["AIRLINE", "AIRLINE_NAME"]].head())
```

```python
# 10-CALCULATING BASIC FLIGHTS KPIs 
#1- TOTAL FLIGHTS 
print("TOTAL FILGHTS:",len(df1))

#2)- CANCELLED FLIGHT
print("CANCELLED FLIGHTS:",df1["CANCELLED"].sum())

#3)-DIVERTED FLIGHTS 
print("DIVERED FLIGHTS:",df1["DIVERTED"].sum())

#4)-AVERAGE DEPARTURE DELAY 
print("AVERAGE DEPARTURE DELAY:",round(df1["DEPARTURE_DELAYED"].mean(), 2), "minutes")

#5- AVERAGE ARRIVAL DELAY 
print("AVERAGE ARRIVAL DELAY:", round (df1["ARRIVAL_DELAYED"].mean(), 2),"minutes")
```

```python
# 11.ANALYZING AIRLINE PERFORMANCE 
airline_analysis=df1.groupby("AIRLINE_NAME").agg(
    Total_Flights=("AIRLINE","size"),
    Cancelled_Flights=("CANCELLED","sum"),
    Average_Departure_Delay=("DEPARTURE_DELAY","mean"),
    Average_Arrival_Delay=("ARRIVAL_DELAY","mean")).reset_index()
print("AIRLINE ANALYSIS:")
print(airline_analysis.head())


# CALCULATING CANCELLATION PERCENTAGE 
airline_analysis["Cancellation_Rate"]=(airline_analysis["Cancelled_Flights"]/airline_analysis["Total_Flights"]*100).round(2)

airline_analysis.sort_values("Average_Arrival_Delay",ascending=False).head(10)
```

```python
# 12.ROUTE ANALYSIS 
route_analysis=df1.groupby("ROUTE").agg(Total_Flights=("ROUTE","size"),
                                        Average_Arrival_Delay=("ARRIVAL_DELAY","mean")
                                       ).reset_index()

route_analysis=route_analysis[route_analysis["Total_Flights"]>=100
    ]
route_analysis.sort_values("Average_Arrival_Delay",ascending=False).head(10)
```

```python
#13. VISUALIZATION 
# 13.1- FLIGHT VOLUME BY MONTH 
print("FLIGHT VOLUME BY MONTHS:")
df1["MONTH"].value_counts().sort_index().plot(kind="bar")
plt.title("Flight Volume by month")
plt.xlabel("Months")
plt.ylabel("Number of flights")
plt.show()
```

```python
#13.2 - CANCELLATION RATE BY AIRLINE 
airline_analysis.sort_values("Cancellation_Rate").plot(x="AIRLINE_NAME",y="Cancellation_Rate",kind="bar",legend=False)

plt.title("Cancellation Rate by Airline")
plt.xlabel("Airline")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#13.3- AVERAGE ARRIVAL DELAY BY SCHEDULE DEPARTURE HOUR
print("AVERAGE ARRIVAL DELAY BY SCHEDULE DEPARTURE HOUR:")
df1["DEPARTURE_HOUR"]=df1["SCHEDULED_DEPARTURE"]//100
df1.loc[df1["DEPARTURE_HOUR"]>23,"DEPARTURE_HOUR"]=np.nan
hour_analysis=df1.groupby("DEPARTURE_HOUR")["ARRIVAL_DELAY"].mean()

hour_analysis.plot(kind="line",marker="o")

plt.title("Average Arrival Delay by Schedule Departure Hour")
plt.xlabel("Schedule Departure Hour")
plt.ylabel("Average Arrival Dealy(Minutes)")
plt.xticks(range(24))
plt.tight_layout()
plt.show()
```
